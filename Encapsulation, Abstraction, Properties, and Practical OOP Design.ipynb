#### Q1. Encapsulation basics
Create a class Customer with attributes name and __balance.
- Initialize both in the constructor.
- Add a method get_balance() that returns the balance.
- Create an object with name "Rahul" and balance 5000.
- Print the customer's name and balance.
Goal: Understand why __balance uses double underscores.

```python
class Customer:
    def __init__(self,name,__balance):
        self.name=name
        self.__balance=__balance

    def get_balance(self):
        return self.__balance
        
c1=Customer('Rahul', 5000)
print(c1.get_balance())
```

##### __balance → triggers name mangling to help avoid accidental name conflicts.

#### Q2. Controlled modification
Create a class BankAccount with a private attribute __balance.
Add methods:
- deposit(amount) — increases the balance if the amount is positive.
- get_balance() — returns the current balance.
Test your class by depositing 1000 into an account that starts with 5000.
Expected balance: 6000.

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance=balance
        
    def deposit(self,amount):
        if amount>0:
            self.__balance+=amount

    def get_balance(self):
        return self.__balance

b=BankAccount(5000)
b.deposit(1000)
print(b.get_balance())
        
```

#### Q3. Data validation

- set_price(price) that updates the price only if it is greater than zero.
- get_price() that returns the current price.

Goal: Protect an object's data from invalid modifications through its public methods.

```python
class Product:
    def __init__(self,price):
        self.__price=price

    def set_price(self,price):
        if price>0:
            self.__price=price

    def get_price(self):
        return self.__price

p=Product(5000)
p.set_price(5500)
print(p.get_price())
p.set_price(-350000)
print(p.get_price())
p.set_price(3500)
print(p.get_price())
        
```

#### Q5. Property setter
Extend the Campaign class with a sessions property.
Requirements:
- Store the value internally as _sessions.
- Return it through a getter.
- Allow updates through a setter.
- Reject negative session counts.
Test the setter by updating the session count from 5000 to 6000, and then try a negative value.

```python
class Campaign:
    def __init__(self,sessions,purchases):
        self._sessions=sessions
        self.purchases=purchases

    @property
    def sessions(self):
        return self._sessions

    @sessions.setter
    def sessions(self,sessions):
        if sessions<0:
            print('sessions cant be negative')
        else:
            self._sessions=sessions

c1=Campaign(350, 150)
c1.sessions
            
c1.sessions=550   
c1.sessions

c1.sessions=-89
```

#### Q6. Static methods
Create a class AnalyticsUtils with a static method calculate_ctr(clicks, impressions).

Test it with 400 clicks and 20,000 impressions.
Expected output: 2.0.
Handle the case where impressions are zero.
Goal: Understand when a method does not need instance attributes or class attributes.

```python
class AnalyticsUtils:
    @staticmethod
    def ctr(clicks,impressions):
        if impressions==0:
            return ('Impressions cant be Zero')
        else:
            return (clicks/impressions)*100

a=AnalyticsUtils()
print(a.ctr(400,20000))
print(a.ctr(400,0))
```

#### Q8. Abstraction
Use Python's abc module to create an abstract class DataLoader.
- Define an abstract method load_data().
- Create a child class CSVLoader that implements load_data() and returns "Loading CSV file".
- Create another child class SQLLoader that implements load_data() and returns "Loading SQL database".
Create objects of both child classes and call their methods.
Challenge: Try creating a DataLoader object directly. What happens, and why?

```python
from abc import ABC, abstractmethod

class DataLoader(ABC):
    @abstractmethod
    def load_data(self):
        pass

class CSVLoader(DataLoader):
    def load_data(self):
        return 'Loading CSV file'

class SQLLoader(DataLoader):
    def load_data(self):
        return 'Loading SQL DB'

c=CSVLoader()
print(c.load_data())
s=SQLLoader()
print(s.load_data())

```

#### Q9. OOP design for data cleaning
Design a class DataCleaner that:
- Accepts a list of numeric values.
- Provides a method to remove None values.
- Provides a method to calculate the mean of the cleaned values.
- Provides a method to return the cleaned dataset.

```python
class DataCleaner:
    def __init__(self,values):
        self.values=values

    def data_cleaner(self):
        clean=[]
        for i in self.values:
            if i is None:
                continue
            else:
                clean.append(i)
        return clean

    def clean_avg(self):
        clean=self.data_cleaner()
        total=0
        for i in clean:
            total+=i
        return total/len(clean)

d=DataCleaner([10, 20, None, 30, 40, None])

print(d.data_cleaner())
print(d.clean_avg())
```

#### Q10. Mini-project — Analytics reporting system
Build a small reusable analytics reporting system using OOP.
Part A — Abstract class AnalyticsReport
- Define an abstract method generate_report().
Part B — Child class CampaignReport
- Accept channel, sessions, purchases, and revenue.
- Implement generate_report().
- Calculate conversion rate and revenue per session.
- Return a readable report containing the channel and both metrics.
Part C — Encapsulation and validation
- Reject negative sessions, purchases, or revenue.
- Handle zero sessions safely.
- Keep report calculations reusable through separate methods or properties.

```python
from abc import ABC, abstractmethod
class AnalyticsReport(ABC):
    @abstractmethod
    def generate_report(self):
        pass
class CampaignReport(AnalyticsReport):
    def generate_report(self):
        print('Campaign Report is being generated!!')

    def validate(self):
        if self.sessions < 0 or self.purchases < 0 or self.revenue < 0:
            raise ValueError("Values cannot be negative")
        if self.purchases > self.sessions:
            raise ValueError("Purchases cannot exceed sessions")


    def __init__(self,channel,sessions,purchases,revenue):
        self.channel=channel
        self.sessions=sessions
        self.purchases=purchases
        self.revenue=revenue

    def metrics(self):
        report={}
        self.validate()
        report['CVR']=(self.purchases/self.sessions)*100
        report['Rev/Session']=(self.revenue/self.sessions)
        self.generate_report()
        return report

report = CampaignReport(
    "Paid Search",
    5000,
    250,
    125000
)    
report=(report.metrics())
print(report)
    

        
```
