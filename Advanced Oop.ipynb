#### Q1. Library Management System
Create a class LibraryBook to manage book information.
Requirements:
- Accept title, author, and total_copies.
- Store the number of available copies in a private attribute __available_copies.
- Create a borrow_book() method that decreases available copies by one.
- Create a return_book() method that increases available copies by one.

#### Q2. Employee Payroll System
Create a base class Employee and two subclasses: FullTimeEmployee and ContractEmployee.
Requirements:
- The base class stores the employee's name and ID.
- Define an abstract method calculate_salary().
- A full-time employee receives a fixed monthly salary plus a bonus.
- A contract employee receives payment based on hours worked and hourly rate.
- Validate salary-related inputs.
- Use polymorphism to calculate salaries for both employee types in the same loop.
Test:

    FullTimeEmployee("Aarav", 101, 50000, 5000),
    ContractEmployee("Meera", 102, 120, 400)
]

for employee in employees:
    print(employee.calculate_salary())

```python
from abc import ABC, abstractmethod
class Employee(ABC):
    def __init__(self,name,ID):
        self.name=name
        self.ID=ID

    @abstractmethod
    def calculate_salary(self):
        print('Salary is being calculated!!')
        pass


class FullTimeEmployee(Employee):
    def __init__(self,name,ID,salary,bonus):
        super().__init__(name,ID)
       
        self.salary=salary
        self.bonus=bonus
        if(self.salary<=0):
            return 'Invalid Input'

    def calculate_salary(self):
        super().calculate_salary()
        return ("Total salary: ",self.salary+self.bonus)


class ContractEmployee(Employee):
    def __init__(self,name,ID,hours,rate):
        super().__init__(name,ID)
        
        self.hours=hours
        self.rate=rate

    
    def calculate_salary(self):
        super().calculate_salary()
        return ("Total salary: ",self.hours*self.rate)

employees = [ FullTimeEmployee("Aarav", 101, 50000, 5000), 
            ContractEmployee("Meera", 102, 120, 400) ]
for employee in employees:
    sal=employee.calculate_salary()
    print(sal)
    
```

#### Q3. E-Commerce Shopping Cart
Create two classes: ProductItem and ShoppingCart.
Requirements:
- ProductItem stores a product name and unit price.
- ShoppingCart manages multiple products and their quantities.
- Use composition: the cart contains product items.
- Implement add_item(), remove_item(), and calculate_total().
- Reject negative prices and non-positive quantities.
- Prevent removing a quantity greater than the quantity in the cart.

cart = ShoppingCart()

cart.add_item(ProductItem("Keyboard", 1500), 2)
cart.add_item(ProductItem("Mouse", 700), 1)

print(cart.calculate_total())  # 3700

```python
class ProductItem:
    def __init__(self,name,price):
        self.name=name
        self.price=price
        if price < 0:
            raise ValueError("Price cannot be negative")
        

class ShoppingCart:
    def __init__(self):
         self.items = {}
   
    def add_item(self, product, quantity):
        if quantity <= 0:
            raise ValueError("Quantity must be positive")

        if product in self.items:
            self.items[product] += quantity
        else:
            self.items[product] = quantity       

    def remove_item(self, product, quantity):
        if quantity <= 0:
            raise ValueError("Quantity must be positive")
        if product not in self.items:
            raise ValueError("Product not found in cart")
        if quantity > self.items[product]:
            raise ValueError("Cannot remove more than cart quantity")

        self.items[product] -= quantity
        if self.items[product] == 0:
            del self.items[product]        
            

    def calculate_total(self):
        total = 0
        for product, quantity in self.items.items():
            total += product.price * quantity

        return total


cart = ShoppingCart()

cart.add_item(ProductItem("Keyboard", 1500), 2)
cart.add_item(ProductItem("Mouse", 700), 1)

print(cart.calculate_total())
    
```

#### Q4. Payment Processing System
Create an abstract class PaymentMethod with an abstract method process_payment().
Create three subclasses:
- CreditCardPayment
- UPIPayment
- WalletPayment
Requirements:
- Each payment type implements process_payment() differently.
- Each accepts a payment amount.
- Reject zero or negative payment amounts.
- Credit-card payments include a processing fee of 2%.
- UPI payments include a fee of 0%.
- Wallet payments include a fee of 1%.
- Return the total amount charged, including the fee.
Test:
payments = [    CreditCardPayment(1000),    UPIPayment(1000),    WalletPayment(1000)]for payment in payments:    print(payment.process_payment())



Expected totals: 1020, 1000, and 1010.

```python
from abc import ABC, abstractmethod

class PaymentMethod(ABC):
    def __init__(self, amount):
        self.amount=amount
        if amount <= 0:
            raise ValueError("Payment amount must be positive")
        
    @abstractmethod
    def process_payment(self):
        pass

class CreditCardPayment(PaymentMethod):
 
    def process_payment(self):
        self.amount=self.amount*1.02
        return self.amount

    
class UPIPayment(PaymentMethod):
    def process_payment(self):
        return self.amount

class WalletPayment(PaymentMethod):
    def process_payment(self):
        self.amount=self.amount*1.01
        return self.amount


payments = [ CreditCardPayment(1000), UPIPayment(1000), WalletPayment(1000)]
for payment in payments:
    print(payment.process_payment())
    
```

#### Q5. Student Grade Management
Create a class StudentRecord.
Requirements:
- Store the student's name and marks in a private attribute __marks.
- Accept marks between 0 and 100.
- Use a property and setter to validate updated marks.
- Create a grade property:
  - 90–100: A
  - 80–89: B
  - 70–79: C
  - 60–69: D
  - Below 60: F
- Create a class attribute passing_marks = 40.
- Add a class method that updates the passing threshold.
- Add a method that returns whether the student has passed.

```python
class StudentRecord:
    passing_marks = 40

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    @property
    def marks(self):
        return self.__marks

    @marks.setter
    def marks(self, marks):
        if 0 <= marks <= 100:
            self.__marks = marks
        else:
            raise ValueError("Marks must be between 0 and 100")

    @property
    def grade(self):
        if self.__marks >= 90:
            return "A"
        elif self.__marks >= 80:
            return "B"
        elif self.__marks >= 70:
            return "C"
        elif self.__marks >= 60:
            return "D"
        else:
            return "F"

    @classmethod
    def update_passing_marks(cls, new_threshold):
        if not 0 <= new_threshold <= 100:
            raise ValueError("Threshold must be between 0 and 100")

        cls.passing_marks = new_threshold

    def has_passed(self):
        return self.__marks >= self.passing_marks

student = StudentRecord("Riya", 86)

print(student.grade)       
print(student.has_passed())

student.marks = 35
print(student.has_passed())
```

- Define an abstract method calculate_rental(days).
- A car receives a 10% discount when rented for at least seven days.

- Validate the number of rental days and the daily rate.
- Use polymorphism to calculate the rental cost of different vehicles.

vehicles = [
    Car("CAR101", 2000),

```python
from abc import ABC, abstractmethod
class Vehicle(ABC):
    def __init__(self,rc,daily_hire_rate):
        if daily_hire_rate <= 0:
            raise ValueError("Daily rental rate must be positive")

        self.rc=rc
        self.daily_hire_rate=daily_hire_rate

    @abstractmethod
    def calculate_rental(self,days):
        pass

class Car(Vehicle):
    def calculate_rental(self,days):
        if days <= 0:
            raise ValueError("Rental days must be positive")
        if days<7:
            return days*self.daily_hire_rate
        else:
            return (days*self.daily_hire_rate)*0.9

class Motorcycle(Vehicle):
    def calculate_rental(self,days):
        if days <= 0:
            raise ValueError("Rental days must be positive")
        if days<5:
            return days*self.daily_hire_rate
        else:
            return (days*self.daily_hire_rate)*0.95

vehicles = [
    Car("CAR101", 2000),
    Motorcycle("BIKE101", 800)
]

for vehicle in vehicles:
    print(vehicle.calculate_rental(7))
            
    
        
```

#### Q7. Inventory Stock Control
Create a class InventoryItem.
Requirements:
- Store the item name, unit price, and stock quantity.
- Keep stock quantity in a private attribute.
- Implement restock(quantity) and sell(quantity).
- Reject negative restocking quantities and non-positive sales quantities.
- Prevent sales that exceed available stock.
- Add a read-only property called inventory_value that calculates stock quantity × unit price.
- Implement __str__() to display the item name, quantity, and inventory value.

```python
class InventoryItem:
    def __init__(self,name,price,quantity):
        if price < 0:
            raise ValueError("Price cannot be negative")
        if quantity < 0:
            raise ValueError("Stock quantity cannot be negative")
        self.name=name
        self.price=price
        self.__quantity=quantity
        
    def restock(self,quantity):
        if quantity<=0:
            raise ValueError('Quantity cant be negative')
        else:
            self.__quantity+=quantity

    def sell(self,quantity):
         if quantity<=0 or quantity>self.__quantity:
            raise ValueError('Quantity out of range')
         else:
            self.__quantity-=quantity
    @property
    def inventory_value(self):
        return self.__quantity * self.price

    def __str__(self):
        return (
            f"Item: {self.name}, "
            f"Quantity Remaining: {self.__quantity}, "
            f"Inventory Value: {self.inventory_value}"
        )


item = InventoryItem("Monitor", 12000, 10)

item.sell(3)
item.restock(5)

print(item.inventory_value)
print(item)

```

#### Q8. Notification System
Create an abstract class Notification with an abstract method send(message).
Create three subclasses:
- EmailNotification
- SMSNotification
- PushNotification
Requirements:
- Each subclass implements send() differently.
- Store the destination address, phone number, or device token in the relevant object.
- Reject empty messages and missing destinations.
- Return a confirmation string rather than printing inside send().
- Demonstrate polymorphism by sending the same message through all three notification types.
Test:

notifications = [
    EmailNotification("student@example.com"),
    SMSNotification("+911234567890"),
    PushNotification("device_token_123")
]

```python
from abc import ABC, abstractmethod
class Notification(ABC):
    def __init__(self, destination):
        if not destination or not destination.strip():
            raise ValueError("Destination cannot be empty")

        self.destination = destination
        
    
    
    @abstractmethod
    def send(self,msg):
        pass

class EmailNotification(Notification):
    def __init__(self, email_address):
        super().__init__(email_address)

    def send(self,msg):
        return f"Email sent to {self.destination}: {msg}"

class SMSNotification(Notification):
    def __init__(self, phone_number):
        super().__init__(phone_number)

    def send(self,msg):
        return f"SMS sent to {self.destination}: {msg}"

class PushNotification(Notification):
    def __init__(self, device_token):
        super().__init__(device_token)

    def send(self,msg):
        return f"PushNotification sent to {self.destination}: {msg}"


notifications = [
    EmailNotification("student@example.com"),
    SMSNotification("+911234567890"),
    PushNotification("device_token_123")
]

for notification in notifications:
    print(notification.send("Your course is starting soon."))
```

#### Q9. Bank Account System
Create a base class BankAccount and two subclasses, SavingsAccount and CurrentAccount.
Requirements:
- Store the account holder and balance privately.
- Implement deposit() and withdraw().
- Reject negative deposits and withdrawals.
- A savings account must maintain a minimum balance of ₹1,000.
- A current account may have an overdraft limit of ₹5,000.
- Use method overriding to enforce the different withdrawal rules.
- Implement __str__() to display the account holder and current balance without exposing internal implementation details.
Test:


accounts = [
    SavingsAccount("Aman", 5000),
    CurrentAccount("Neha", 2000)
]

for account in accounts:
    account.withdraw(3000)
    print(account)

```python

class BankAccount:
    def __init__(self, account_holder, balance):
        if not account_holder or not account_holder.strip():
            raise ValueError("Account holder name cannot be empty")

        if balance < 0:
            raise ValueError("Initial balance cannot be negative")

        self.__account_holder = account_holder
        self.__balance = balance

    @property
    def balance(self):
        return self.__balance

    @property
    def account_holder(self):
        return self.__account_holder

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit amount must be positive")

        self.__balance += amount
        return self.__balance

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Withdrawal amount must be positive")

        if amount > self.__balance:
            raise ValueError("Insufficient balance")

        self.__balance -= amount
        return self.__balance

    def _set_balance(self, new_balance):
        self.__balance = new_balance

    def __str__(self):
        return (
            f"Account Holder: {self.__account_holder}, "
            f"Balance: ₹{self.__balance:,.2f}"
        )


class SavingsAccount(BankAccount):
    MINIMUM_BALANCE = 1000

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Withdrawal amount must be positive")

        if self.balance - amount < self.MINIMUM_BALANCE:
            raise ValueError(
                "Withdrawal denied: minimum balance of ₹1,000 must be maintained"
            )

        self._set_balance(self.balance - amount)
        return self.balance


class CurrentAccount(BankAccount):
    OVERDRAFT_LIMIT = 5000

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Withdrawal amount must be positive")

        if self.balance - amount < -self.OVERDRAFT_LIMIT:
            raise ValueError("Withdrawal denied: overdraft limit exceeded")

        self._set_balance(self.balance - amount)
        return self.balance



accounts = [
    SavingsAccount("Aman", 5000),
    CurrentAccount("Neha", 2000)
]

for account in accounts:
    try:
        account.withdraw(3000)
    except ValueError as error:
        print(error)

    print(account)
```

#### Q10. Online Course Enrollment Platform
Build a mini-project using several OOP concepts together.
Create these classes:
- Course: stores course title, fee, and maximum capacity.
- Student: stores student information and enrolled courses.
- OnlineCourse: inherits from Course and adds an access duration.
- EnrollmentManager: manages enrollment and course capacity.
Requirements:
1. Encapsulate the course's enrolled-student count.
2. Prevent enrollment when the course is full.
3. Prevent the same student from enrolling in the same course twice.
4. Use composition to let EnrollmentManager manage courses and students.
5. Use inheritance to extend the basic course with OnlineCourse.
6. Use a property to calculate remaining seats.
7. Validate course fees, capacity, and access duration.
8. Implement __str__() for readable course information.
9. Add a class method to track the total number of course objects created.
10. Demonstrate enrolling students, handling errors, and displaying remaining seats.
Example starting point:


class Course:

    total_courses = 0

    def __init__(self, title, fee, capacity):
        # Initialize and validate course data
        pass
Your goal is to build a working system, not just individual methods.

```python

class Course:
    total_courses = 0

    def __init__(self, title, fee, max_capacity):
        if not title or not title.strip():
            raise ValueError("Course title cannot be empty")

        if fee < 0:
            raise ValueError("Course fee cannot be negative")

        if max_capacity <= 0:
            raise ValueError("Course capacity must be positive")

        self.title = title
        self.fee = fee
        self.max_capacity = max_capacity
        self.__enrolled_count = 0
        self._enrolled_students = set()

        Course.total_courses += 1

    @property
    def remaining_seats(self):
        return self.max_capacity - self.__enrolled_count

    @property
    def enrolled_count(self):
        return self.__enrolled_count

    def _enroll_student(self, student):
        if student in self._enrolled_students:
            raise ValueError(
                f"{student.name} is already enrolled in {self.title}"
            )

        if self.remaining_seats <= 0:
            raise ValueError(f"{self.title} is full")

        self._enrolled_students.add(student)
        self.__enrolled_count += 1

    def __str__(self):
        return (
            f"Course: {self.title} | "
            f"Fee: ₹{self.fee:,.2f} | "
            f"Capacity: {self.max_capacity} | "
            f"Enrolled: {self.__enrolled_count} | "
            f"Remaining seats: {self.remaining_seats}"
        )

    @classmethod
    def get_total_courses(cls):
        return Course.total_courses


class OnlineCourse(Course):
    def __init__(self, title, fee, max_capacity, access_duration):
        if access_duration <= 0:
            raise ValueError("Access duration must be positive")

        super().__init__(title, fee, max_capacity)
        self.access_duration = access_duration

    def __str__(self):
        return (
            f"{super().__str__()} | "
            f"Access duration: {self.access_duration} months"
        )


class Student:
    def __init__(self, student_id, name):
        if not student_id or not student_id.strip():
            raise ValueError("Student ID cannot be empty")

        if not name or not name.strip():
            raise ValueError("Student name cannot be empty")

        self.student_id = student_id
        self.name = name
        self.__enrolled_courses = set()

    @property
    def enrolled_courses(self):
        return tuple(self.__enrolled_courses)

    def _add_course(self, course):
        self.__enrolled_courses.add(course)


class EnrollmentManager:
    def __init__(self):
        self.courses = {}
        self.students = {}

    def add_course(self, course):
        if course.title in self.courses:
            raise ValueError("A course with this title already exists")

        self.courses[course.title] = course

    def add_student(self, student):
        if student.student_id in self.students:
            raise ValueError("Student ID already exists")

        self.students[student.student_id] = student

    def enroll(self, student_id, course_title):
        if student_id not in self.students:
            raise ValueError("Student not found")

        if course_title not in self.courses:
            raise ValueError("Course not found")

        student = self.students[student_id]
        course = self.courses[course_title]

        # Validate and update the course first.
        course._enroll_student(student)

        # Update the student's enrollment record.
        student._add_course(course)

        return f"{student.name} enrolled in {course.title} successfully"




manager = EnrollmentManager()

python_course = OnlineCourse(
    "Python OOP",
    2500,
    2,
    6
)

sql_course = Course(
    "SQL Analytics",
    1800,
    3
)

manager.add_course(python_course)
manager.add_course(sql_course)

aman = Student("S101", "Aman")
neha = Student("S102", "Neha")
rahul = Student("S103", "Rahul")

manager.add_student(aman)
manager.add_student(neha)
manager.add_student(rahul)

# Successful enrollments
print(manager.enroll("S101", "Python OOP"))
print(manager.enroll("S102", "Python OOP"))

# Duplicate enrollment
try:
    manager.enroll("S101", "Python OOP")
except ValueError as error:
    print("Error:", error)

# Course is full
try:
    manager.enroll("S103", "Python OOP")
except ValueError as error:
    print("Error:", error)

# Display course information
print()
print(python_course)
print(sql_course)

print("\nTotal course objects created:", Course.get_total_courses())
print("Aman's enrolled courses:", [c.title for c in aman.enrolled_courses])
```
