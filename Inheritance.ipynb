# Basic Inheritance

#### Q1. Basic inheritance
Create a parent class Campaign with:
- Constructor accepting channel and sessions.
- Method show_channel() returning the channel name.
Create a child class PaidCampaign that inherits from Campaign and adds:
- Attribute spend.
- Method cost_per_session() returning spend / sessions.
Input: channel = "Paid Search"
sessions = 5000
spend = 10000

```python
class Campaign:
    def __init__(self,channel,sessions):
        self.channel=channel
        self.sessions=sessions

    def show_channel(self):
        return self.channel

class PaidCampaign(Campaign):
    def __init__(self,channel,sessions,spend):
        super().__init__(channel,sessions)
        self.spend=spend

    def cost_per_session(self):
        return self.spend/self.sessions

p1=PaidCampaign('paid search',5000,10000)
print(p1.show_channel(),p1.cost_per_session())
```

#### Q2. Inheritance and constructor

Create an object with:

accuracy = 0.92
model_type = "Classification"

```python
class Model:
    def __init__(self,name,accuracy):
        self.name=name
        self.accuracy=accuracy

class ClassificationModel(Model):
    def __init__(self,name,accuracy,model_type):
        super().__init__(name,accuracy)
        self.model_type=model_type

m1= ClassificationModel('Logistic Regression', 0.92,'Classification')
print(m1.name,m1.accuracy,m1.model_type)
```

#### Q4. Method overriding

- ClassificationModel, returning "Predicting a category".
- RegressionModel, returning "Predicting a numeric value".

#### Q5. Combining inheritance and calculations
Create a parent class Campaign with attributes channel, sessions, and purchases.
Add methods to calculate:
- Conversion rate: purchases / sessions * 100
- A summary of the campaign.

Create a child class RevenueCampaign that adds revenue and a method to calculate revenue per session.
Input:
channel = "Email"sessions = 2000purchases = 220revenue = 110000



Calculate the conversion rate and revenue per session.
Expected calculations: Conversion rate = 11.0%; revenue per session = 55.0.

```python
class Campaign:
    def __init__(self,channel,sessions,purchases):
        self.channel=channel
        self.sessions=sessions
        self.purchases=purchases

    def cvr(self):
        return (self.purchases/self.sessions)*100

    def summary(self):
        return (
        f"Channel: {self.channel}\n"
        f"Sessions: {self.sessions}\n"
        f"Purchases: {self.purchases}\n"
        f"Conversion Rate: {self.cvr():.1f}%"
    )
        

class RevenueCampaign(Campaign):
    def __init__(self,channel,sessions,purchases,revenue):
        super().__init__(channel,sessions,purchases)
        self.revenue=revenue

    def rev_per_session(self):
        return (
        f'\nRevenue per session: {self.revenue/self.sessions}')

rc1=RevenueCampaign('Email',2000,220,110000)
print(rc1.summary(),rc1.rev_per_session())
```

#### Q7. Duck typing
Create two classes:
- CSVData, with a method load() returning "Loading CSV data".
- SQLData, with a method load() returning "Loading SQL data".
Write a function load_dataset(source) that calls source.load() and returns the result.
Test the function with objects of both classes.
Question: Why can the same function work with two different classes without requiring them to share a parent class?\


Answer:
if two different classes both have a load() method, the same function can work with objects of either class without requiring them to share a parent class. this is called duck typing

```python
class CSVData:
    def load(self):
        return 'Loading CSV Data'

class SQLData:
    def load(self):
        return 'Loading SQL data'

def load_data(source):
    print(source.load())

c=CSVData()
s=SQLData()
load_data(c)
load_data(s)
```

#### Q8. Inheritance with data cleaning
Create a parent class DataProcessor with:
- A constructor accepting a list of values.
- A method remove_missing() that removes None values.
Create a child class NumericProcessor with a method calculate_mean() that calculates the mean of the cleaned numeric values.

Input: values = [10, 20, None, 30, 40, None]

```python
class DataProcessor:
    def __init__(self,values):
        self.values=values

    def remove_missing(self):
        clean=[]
        for i in self.values:
            if i is None:
                continue
            else:
                clean.append(i)
        return clean

class NumericProcessor(DataProcessor):
    def __init__(self,values):
        super().__init__(values)

    def calculate_mean(self):
        val=self.remove_missing()
        total=0
        for i in val:
            total+=i
        print('mean of values: ',total/len(val))

n=NumericProcessor( [10, 20, None, 30, 40, None])
n.calculate_mean()
    
    
    
```

#### Q9. Method overriding with super()
Create a parent class WebReport with a method generate_report() returning:
Basic web analytics report


Create a child class CampaignReport that overrides generate_report().
Use super() to retrieve the parent's report and append:
Campaign conversion analysis included


Expected output:
Basic web analytics report
Campaign conversion analysis included

```python
class WebReport:
    def generate_report(self):
        return 'Basic web analytics report'

class CampaignReport(WebReport):
    def generate_report(self):
        report=super().generate_report()+' Campaign conversion analysis included'
        return report

c=CampaignReport()
c.generate_report()

    
```

#### Q10. Mini-project — ML model evaluation system
Build a small object-oriented system.
Parent class: MLModel
- Constructor: name, accuracy.
- Method model_summary() returning the model name and accuracy.
Child class: ClassificationModel
- Inherits from MLModel.
- Adds precision and recall.
- Overrides model_summary() to include the inherited summary and both additional metrics, using super().
Child class: RegressionModel
- Inherits from MLModel.
- Adds mae (mean absolute error).
- Overrides model_summary() to include the inherited summary and MAE, using super().

```python
class MLModel:
    def __init__(self,name,accuracy):
        self.name=name
        self.accuracy=accuracy

    def model_summary(self):
        return (self.name,self.accuracy)

class ClassificationModel(MLModel):
    def __init__(self,name,accuracy,precision,recall):
        super().__init__(name,accuracy)
        self.precision=precision
        self.recall=recall

    def model_summary(self):
        print(super().model_summary(),self.precision,self.recall)

class RegressionModel(MLModel):
    def __init__(self,name,accuracy,mae):
        super().__init__(name,accuracy)
        self.mae=mae

    def model_summary(self):
         print(super().model_summary(),self.mae)

classification = ClassificationModel(
    "Logistic Regression",
    0.92,
    0.90,
    0.88
)
classification.model_summary()

regression = RegressionModel(
    "Linear Regression",
    0.85,
    12.5
)    
regression.model_summary()
        
```
