#### Q1. Instance attributes vs. class attributes
Create an Employee class with a class attribute company = "DataTech".
Initialize name and salary as instance attributes.
Create two employees with different names and salaries. Print their names, salaries, and company.

```python
class Employee:
    company='DataTech'
    def __init__(self,name,salary):
        self.name=name
        self.salary=salary

e1=Employee('Shorya',1500)
e2=Employee('Rajiv',1800)
print(e1.company,e1.name,e1.salary)
print(e2.company,e2.name,e2.salary)
```

#### Q6. Create a class method
Create an Employee class with a class attribute company = "DataTech".
Add a class method called change_company(cls, new_company) that updates the company name.
Create two employee objects, call the class method to change the company to "Microsoft", and print the company name for both objects.

```python
class Employee:
    company='DataTech'
    @classmethod
    def change_company(cls, new_company):
        cls.company=new_company

e1=Employee()
e2=Employee()
Employee.change_company('Microsoft')
print(e1.company)
print(e2.company)
```

#### Q8. Create a static method
Create a Dataset class with a static method is_valid_value(value).
The method should return True if the value is an integer or float and is not a Boolean; otherwise, return False.
Test it with:
103.5"20"True



Hint: In Python, bool is a subclass of int, so consider using isinstance() carefully.

```python
class Dataset:
    @staticmethod
    def is_valid_value(value):
        if isinstance(value,(int,float)) and not isinstance(value, bool):
            return True
        else:
            return False

print(Dataset.is_valid_value(103.5))
print(Dataset.is_valid_value('20'))

print(Dataset.is_valid_value(False))
```

#### Q9. Class method vs. static method
Create a DiscountCalculator class with a class attribute default_discount = 10.
Implement:
- set_discount(cls, discount) — a class method that changes the default discount.
- calculate_price(price, discount) — a static method that returns the discounted price.
Test the methods using a product priced at ₹2,000. Change the default discount to 15% and calculate the discounted price.

```python
class DiscountCalculator:
    default_discount = 10
    @classmethod
    def set_discounts(cls,discount):
        cls.default_discount=discount

    @staticmethod
    def calculate_price(price, discount):
        return (price-price*(discount/100))

d1=DiscountCalculator()
d1.set_discounts(15)
print(DiscountCalculator.calculate_price(2000,15))
    
    
```
