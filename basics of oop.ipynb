#### Q5. Add an instance method
Create a Circle class that stores radius.
Add an area() method that returns the area using this formula:

```python
class Circle:
    def __init__(self,radius):
        self.radius=radius

    def area(self):
        return 3.14*(self.radius**2)


c1=Circle(5)
print(c1.area())
```

#### Q6. Add a calculation method
Create a Product class with name and price attributes.
Add a discounted_price() method that returns the price after a 10% discount.
Example: If the price is ₹1,000, the discounted price should be ₹900.

```python
class Product:
    def __init__(self,name,price):
        self.name=name
        self.price=price

    def discounted_price(self):
        return self.price*0.9

p1=Product('phone',10000)
print(p1.discounted_price())
```

#### Q9. Build a data-science class
Create a Dataset class initialized with a list of numbers.
Add two methods:
- mean() — returns the average of the numbers.
- count() — returns the number of observations.
Handle an empty list appropriately so that calculating the mean does not cause a division-by-zero error.

```python
class Dataset:
    def __init__(self,data):
        self.data=data

    def count(self):
        if len(self.data)==0:
            print("Empty List")
        return len(self.data)

    def mean(self):
        total = 0
        for i in self.data:
            total += i
        return total / len(self.data)
        


d1=Dataset([1,2,3,4,5,6])
print(d1.count())
print(d1.mean())    
```

#### Q10. Mini-project: Campaign analysis
Create a Campaign class with these attributes:
- channel
- sessions
- purchases
- 
Add two methods:
1. conversion_rate() — calculates the conversion rate as a percentage
2. summary() — returns a summary containing the channel, sessions, purchases, and conversion rate.

Use this sample data:
channel = "Paid Search" sessions = 5000 purchases = 250



Expected conversion rate: 5.0%

```python
class Campaign:
    def __init__(self,channel,sessions,purchases):
        self.channel=channel
        self.sessions=sessions
        self.purchases=purchases

    def cvr(self):
        return (self.purchases/self.sessions)*100

    def summary(self):
        report={}
        report['channel']=self.channel
        report['sessions']=self.sessions
        report['purchases']=self.purchases
        report['cvr']=(self.purchases/self.sessions)*100
        return report

c1=Campaign('Paid search',5000,250)
print(c1.cvr())
report=c1.summary()

print(report)
        
            
```
